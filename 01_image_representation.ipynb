# Create a 2D image array and print it

```python
import numpy as np

image = np.array([
    [0,   50,  100],
    [150, 200, 255],
    [30,  80,  220]
], dtype=np.uint8)

print(image)
```

# Inspect image properties (shape, dtype, min, max)

```python
print("shape:", image.shape)
print("dtype:", image.dtype)
print("min:", image.min())
print("max:", image.max())
```

# Create an RGB image array and print it

```python
rgb_image = np.array([
    [[255,   0,   0], [  0, 255,   0]],
    [[  0,   0, 255], [255, 255, 255]]
], dtype=np.uint8)

print(rgb_image)
```

# Inspect RGB image properties

```python
print("shape:", rgb_image.shape)
print("dtype:", rgb_image.dtype)
print("min:", rgb_image.min())
print("max:", rgb_image.max())
```

# Extract the red channel from the RGB image

```python
print(rgb_image[:, :, 0])
```

# Demonstrate data types and arithmetic operations on bounded integers

```python
a = np.array([250], dtype=np.float32)
b = np.array([10], dtype=np.float32)

result = a + b

print("a:", a)
print("b:", b)
print("result:", result)
print("result dtype:", result.dtype)
```

dtype determines how numerical image values are represented, and arithmetic on bounded integer types can produce overflow.

# Create an image using arange and reshape

```python
image = np.arange(20).reshape(4, 5)

print(image)
print(image.shape)
```

# Access a single pixel

```python
pixel = image[2, 3]
print(pixel)
```

# Slicing rows, columns, and regions

```python
row = image[1, :]
column = image[:, 2]
region = image[1:3, 1:4]

print(row)
print(column)
print(region)
```

# Initialize an HWC image array

```python
image_hwc = np.zeros((2, 3, 3), dtype=np.uint8)

image_hwc[:, :, 0] = 10
image_hwc[:, :, 1] = 50
image_hwc[:, :, 2] = 100
```

# Transpose HWC to CHW format

```python
image_chw = np.transpose(image_hwc, (2, 0, 1))
```

# Print shapes after transposition

```python
print(image_hwc.shape)
print(image_chw.shape)
```

# Verify array equality between HWC channel and CHW slice

```python
np.array_equal(
    image_hwc[:, :, 0],
    image_chw[0, :, :]
)
```
